```python
# Asking the user for weight in Pounds(lb)
weight = int(input('Enter your weight in pounds : '))

# Asking the user for height in inches
height = int(input('Enter your height in inches : '))

# Calculating the BMI with the formula (weight(lb) x 703)/(height x height)
BMI = round((weight * 703)/ (height * height) ,2)

# Calculaing the BMI band according to BMI calculated above
print('\n')
if(BMI > 0): 
    if(BMI < 18.5):
        print('You are Underweight with your BMI of : ', BMI)
    elif(BMI <= 24.9):
        print('You are Healthy with your BMI of : ', BMI)
    elif(BMI <= 29.9):
        print('Your are Overweight with your BMI of : ', BMI)
    elif(BMI <= 34.9):
        print('You are Obese with your BMI of : ', BMI)
    elif (BMI <= 39.9):
        print('You are in High Obesity Category with your BMI of : ', BMI)
    elif(BMI >= 40):
        print('You are in Extreme Obesity Category with your BMI of : ', BMI)
else:
    print('Please enter valid inputs.')
```
