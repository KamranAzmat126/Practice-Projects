```python
import os, shutil
```

```python
path = r"D:/Data Analyst Course by Alex/Python/File Sorter Project/"
```

```python
files = os.listdir(path)
print(files)
```

```python
for file in files:
    if(os.path.isfile(path+file)):
        if(os.path.isdir(path + (file.rsplit('.',1)[1].upper() + ' Files'))):
            shutil.move(path+file, path + ((file.rsplit('.',1)[1]).upper() + ' Files'))
        else:
            os.makedirs(path + (file.rsplit('.',1)[1]).upper() + ' Files')
            shutil.move(path+file, path + ((file.rsplit('.',1)[1]).upper() + ' Files'))
    else:
        print('No File Existed!')
```
